<h2>Part III: Conditional Subsetting</h2>
In plain language: "Give me all of the diamonds that have a carat weight greater than two." 

```python
# diamonds where carat is greater than 2
diamonds[ (diamonds['carat'] > 2) ]
```

We can add more conditions to the end of the line.

```python
# filtering feature qualities
diamonds[ (diamonds['carat']   <= 1) ]\
        [ (diamonds['color']   >= 2) ]\
        [ (diamonds['clarity'] == 3) ]\
        [ (diamonds['cut']     != 1) ]
```

This gets messy fast. It is easier to build the code one step at a time.

```python
# data to be subset
data = diamonds[ ['carat', 'color', 'clarity', 'cut'] ]


# conditions
carat_condition   = (diamonds['carat']   <= 1) # carat
color_condition   = (diamonds['color']   >= 2) # color
clarity_condition = (diamonds['clarity'] == 3) # clarity
cut_condition     = (diamonds['cut']     != 1) # cut


# putting it all together
data[carat_condition][color_condition][clarity_condition][cut_condition]
```

<strong>Cleaner alternative:</strong> join the conditions with <code>&</code> (and) inside one bracket.
This gives the same result without the reindex warning.

```python
# same filter, combined with & in one step
data[ carat_condition & color_condition & clarity_condition & cut_condition ]
```

<h2>Part IV: Case Study - Finding the "Best" Diamond on a Budget</h2>
<strong>Goal: find the "best" five diamond engagement rings for a "normal" Chicago citizen.</strong>
<hr style="height:.9px;border:none;color:#333;background-color:#333;" />
<strong>Foundational Domain Knowledge</strong><br>
In 2007, per capita income in Chicago was about \$34,775 (U.S. Census Bureau).
Under the "two month's salary" rule, the budget is about <strong>\$5,795</strong> (34,775 / 12 × 2).
<hr style="height:.9px;border:none;color:#333;background-color:#333;" />

```python
# calculating the budget from the two month's salary rule
income = 34775                    # Chicago per capita income, 2007
budget = round(income / 12 * 2)   # two months of salary
print(f"Budget: ${budget:,}")

# how many diamonds can the buyer afford at all?
affordable = diamonds[ (diamonds['price'] <= budget) ]
print(f"Diamonds within budget: {len(affordable)} out of {len(diamonds)}")
```

<h3>Rationale for the selection criteria</h3>

* **Price ≤ \$5,795.** The budget is a hard limit. I treat it as the most the buyer will spend, before tax.
* **Carat ≥ 0.9.** Size is the first thing most people notice in a ring. Prices jump once a diamond reaches 1 carat, so stones from 0.9 to 1.0 carat look almost the same size but cost less. A 0.9 floor keeps those good-value stones in the search.
* **Color ≤ 4 (D to G).** Lower numbers are better. Grades D to G are "colorless" or "near colorless", so the stone will not look yellow to a normal person.
* **Clarity ≤ 5 (VS2 or better).** Lower numbers are better. At VS2 and above, flaws cannot be seen without magnification, so paying for higher clarity adds little visible value.
* **Cut is not filtered.** The documentation says the cut measure is not reliable, so I do not use it now. The filter stays in the code (`<= 1` keeps both values) so I can change it if better data appears.
* **Ranking by price per carat.** To rank the final stones, I compute price ÷ carat. A lower number means the buyer gets more diamond for each dollar. I sort by carat (largest first) and then by price per carat (cheapest first).

These rules mix size, look, and value. They are reasonable for a buyer who is not a diamond expert and wants the most visible quality for the money.

```python
# filters developed for each descriptive metric
price_condition   = (diamonds['price']   <= budget)  # within the two month's salary budget
carat_condition   = (diamonds['carat']   >= 0.9  )   # close to 1 carat or bigger
color_condition   = (diamonds['color']   <= 4    )   # D to G: colorless / near colorless
clarity_condition = (diamonds['clarity'] <= 5    )   # VS2 or better: no visible flaws
cut_condition     = (diamonds['cut']     <= 1    )   # cut not trusted, so keep both values


# applying all conditions at once
best = diamonds[ price_condition   &
                 carat_condition   &
                 color_condition   &
                 clarity_condition &
                 cut_condition ].copy()


# value metric: dollars paid per carat (lower = more diamond per dollar)
best['price_per_carat'] = (best['price'] / best['carat']).round(0)


# ranking: biggest stone first, then best value
best = best.sort_values(by        = ['carat', 'price_per_carat'],
                        ascending = [False,   True])


# displaying the top five
best[ ['carat', 'color', 'clarity', 'cut', 'channel', 'store', 'price', 'price_per_carat'] ].head(n = 5)
```

```python
# same result written in the course's chained style
diamonds[ (diamonds['price']   <= 5795) ]\
        [ (diamonds['carat']   >= 0.9 ) ]\
        [ (diamonds['color']   <= 4   ) ]\
        [ (diamonds['clarity'] <= 5   ) ]\
        [ (diamonds['cut']     <= 1   ) ]
```

```python
# comparing the top five with the whole market
print(f"Average price per carat, all diamonds : ${(diamonds['price'] / diamonds['carat']).mean():,.0f}")
print(f"Average price per carat, top five     : ${best['price_per_carat'].head(5).mean():,.0f}")
print(f"Average spend of top five             : ${best['price'].head(5).mean():,.0f} (budget ${budget:,})")
```

<h3>Conclusion</h3>

Exactly <strong>five diamonds</strong> pass every filter, and all of them are within the \$5,795 budget.

* **Top pick: Obs 81.** 1.14 carats, color 4 (G), clarity 4 (VS1), \$5,472. It is the biggest stone in the group, it has the best clarity, and it has the lowest price per carat (about \$4,800). It also leaves about \$320 of the budget unused.
* **Best color: Obs 83 (1.01 ct, color 2 = E) and Obs 249 (0.91 ct, color 1 = D).** These suit a buyer who cares most about a perfectly white stone.
* **Cheapest: Obs 231.** 0.91 carats, color 2 (E), \$4,841. It saves almost \$1,000 against the budget.
* **Channel:** two of the five are sold in malls (channel 1) and three online (channel 2). None come from an independent jeweler, so a careful buyer should look at mall and internet sellers.

If more information on cut quality becomes available, the `cut_condition` can be changed to `== 1` to keep only Ideal cuts.
